## 9. Convolutional Autoencoder Design

The autoencoder will be composed of two parts:

### Encoder

The encoder receives an image of size **96 × 96 × 3** and progressively reduces its spatial dimensions while increasing the number of feature maps,the goal is to obtain a compact latent representation.the encoder should learn useful visual features such as edges, textures, shapes, and object parts...

### Decoder

The decoder receives the latent representation and reconstructs an image with the same size as the original input,the decoder should progressively increase the spatial resolution until reaching the original image size.
